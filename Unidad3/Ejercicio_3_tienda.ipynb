```python
import numpy as np
import matplotlib.pyplot as plt

# Datos de la distribución de probabilidad (Unidades vendidas y Probabilidades)
X = np.array([0, 1, 2, 3, 4, 5])
P_X = np.array([0.05, 0.10, 0.25, 0.30, 0.20, 0.10])

print("--- RESULTADOS DEL EJERCICIO 3 ---")

# 1. Verificar que las probabilidades sumen 1
suma_prob = np.sum(P_X)
print(f"1. Suma de las probabilidades: {suma_prob:.2f}")

# 3. Calcular Valor esperado, Varianza y Desviación estándar
# Valor esperado E(X) = Promedio de ventas
E_X = np.sum(X * P_X)
print(f"\n3. Valor esperado E(X): {E_X:.2f} unidades")

# Varianza y Desviación Estándar
varianza = np.sum(((X - E_X)**2) * P_X)
print(f"   Varianza: {varianza:.2f}")

desviacion_est = np.sqrt(varianza)
print(f"   Desviación estándar: {desviacion_est:.2f} unidades")

# 4. Construir la función de distribución acumulada F(x)
F_X = np.cumsum(P_X)
print("\n4. Función de distribución acumulada F(x):")
for x_val, fx_val in zip(X, F_X):
    print(f"   F({x_val}) = {fx_val:.2f}")

# 5. Interpretación de los resultados
print("\n5. Interpretación de los resultados:")
print(f"   - Unidades esperadas: Se espera vender en promedio {E_X:.1f} unidades por semana.")
print(f"   - Estabilidad de la demanda: La demanda tiene una variabilidad moderada/baja.")
print(f"     Con una desviación estándar de {desviacion_est:.1f} unidades respecto a la media ({E_X:.1f}),")
print(f"     lo más probable es que las ventas semanales fluctúen entre 2 y 4 unidades.")

# Representación gráfica de F(x)
plt.figure(figsize=(8, 5))
plt.step(X, F_X, where='post', color='orange', linewidth=2, marker='o')
plt.title('Función de Distribución Acumulada F(x) - Ventas Semanales')
plt.xlabel('Unidades vendidas (X)')
plt.ylabel('Probabilidad Acumulada F(x)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(X)
plt.yticks(np.arange(0, 1.1, 0.1))
plt.ylim(0, 1.05)
plt.show()
```
