```python
import numpy as np
import matplotlib.pyplot as plt

# Datos de la distribución de probabilidad obtenidos de la tabla
X = np.array([0, 1, 2, 3, 4, 5])
P_X = np.array([0.10, 0.25, 0.30, 0.20, 0.10, 0.05])

print("--- RESULTADOS DEL EJERCICIO ---")

# 1. Verificar que las probabilidades sumen 1
suma_prob = np.sum(P_X)
print(f"1. Suma de las probabilidades: {suma_prob:.2f}")

# 3. Calcular Valor esperado, Varianza y Desviación estándar
E_X = np.sum(X * P_X)
varianza = np.sum(((X - E_X)**2) * P_X)
desviacion_est = np.sqrt(varianza)

print(f"\n3. Valor esperado E(X): {E_X:.2f} células anormales")
print(f"   Varianza: {varianza:.2f}")
print(f"   Desviación estándar: {desviacion_est:.2f} células")

# 4. Construir la función de distribución acumulada F(x)
F_X = np.cumsum(P_X)
print("\n4. Función de distribución acumulada F(x):")
for x_val, fx_val in zip(X, F_X):
    print(f"   F({x_val}) = {fx_val:.2f}")

# 5. Interpretación en el contexto biomédico
print("\n5. Interpretación en contexto biomédico:")
print(f"   - Promedio esperado: En promedio, se espera encontrar {E_X:.1f} células anormales por muestra de tejido.")
print(f"   - Variabilidad: Existe una variabilidad moderada con una desviación estándar de {desviacion_est:.1f} células.")
print("     Esto indica que la gran mayoría de las muestras analizadas presentarán entre 1 y 3 células anormales.")
print("   - Función acumulada: F(x) indica la probabilidad de encontrar 'hasta' una cierta cantidad de células anormales.")
print(f"     Por ejemplo, F(2) = {F_X[2]:.2f} significa que existe un {F_X[2]*100:.0f}% de probabilidad de encontrar")
print("     2 células anormales o menos en una muestra al azar. Asimismo, muestra que es poco probable encontrar")
print(f"     números altos, ya que hay un {F_X[4]*100:.0f}% de probabilidad de tener 4 o menos (F(4) = {F_X[4]:.2f}).")

# Representación gráfica de F(x)
plt.figure(figsize=(8, 5))
plt.step(X, F_X, where='post', color='blue', linewidth=2, marker='o')
plt.title('Función de Distribución Acumulada F(x) de Células Anormales')
plt.xlabel('Número de células anormales (X)')
plt.ylabel('Probabilidad Acumulada F(x)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(X)
plt.yticks(np.arange(0, 1.1, 0.1))
plt.ylim(0, 1.05)
plt.show()
```
