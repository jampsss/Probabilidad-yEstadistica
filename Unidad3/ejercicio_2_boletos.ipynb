```python
import numpy as np
import matplotlib.pyplot as plt

# Datos de la tómbola obtenidos de la tabla
X = np.array([0, 10, 20, 50])
P_X = np.array([0.50, 0.30, 0.15, 0.05])

print("--- RESULTADOS DEL EJERCICIO 2 ---")

# 1. Verificar que las probabilidades sumen 1
suma_prob = np.sum(P_X)
print(f"1. Suma de las probabilidades: {suma_prob:.2f}")

# 3. Calcular Valor esperado, Varianza y Desviación estándar
E_X = np.sum(X * P_X)
varianza = np.sum(((X - E_X)**2) * P_X)
desviacion_est = np.sqrt(varianza)

print(f"\n3. Valor esperado E(X): ${E_X:.2f} pesos")
print(f"   Varianza: {varianza:.2f}")
print(f"   Desviación estándar: ${desviacion_est:.2f} pesos")

# 4. Construir la función de distribución acumulada F(x)
F_X = np.cumsum(P_X)
print("\n4. Función de distribución acumulada F(x):")
for x_val, fx_val in zip(X, F_X):
    print(f"   F({x_val}) = {fx_val:.2f}")

# 5. Interpretación de los resultados
print("\n5. Interpretación de los resultados:")
print(f"   - Significado del valor esperado: A largo plazo, la ganancia promedio por boleto es de ${E_X:.2f} pesos.")
print(f"   - ¿Es favorable el juego?: Depende del costo del boleto (dato no proporcionado).")
print(f"     Si el boleto cuesta más de ${E_X:.2f} pesos, el juego NO es favorable para el jugador,")
print("     ya que en promedio perdería dinero. Si cuesta menos, sí sería favorable.")

# Representación gráfica de F(x)
plt.figure(figsize=(8, 5))
plt.step(X, F_X, where='post', color='green', linewidth=2, marker='o')
plt.title('Función de Distribución Acumulada F(x) de los Premios')
plt.xlabel('Monto del premio en pesos (X)')
plt.ylabel('Probabilidad Acumulada F(x)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.xticks(X) # Asegura que solo se muestren los valores reales de los premios en el eje X
plt.yticks(np.arange(0, 1.1, 0.1))
plt.ylim(0, 1.05)
plt.show()
```
