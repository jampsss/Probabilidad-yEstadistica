```python
import math

# Definir los parámetros del problema
n = 10  # Total de asistentes disponibles
r = 4   # Preguntas a calificar (asistentes a seleccionar)

# Método 1: Cálculo paso a paso usando factoriales
factorial_n = math.factorial(n)
factorial_resta = math.factorial(n - r)
permutaciones_paso_a_paso = factorial_n // factorial_resta

# Método 2: Función directa de permutaciones de math
permutaciones_directa = math.perm(n, r)

# Mostrar resultados
print(f"Total de asistentes (n): {n}")
print(f"Preguntas a calificar (r): {r}")
print("---")
print(f"10! = {factorial_n}")
print(f"(10 - 4)! = 6! = {factorial_resta}")
print(f"Resultado paso a paso ({factorial_n} / {factorial_resta}): {permutaciones_paso_a_paso}")
print(f"Resultado directo con math.perm(10, 4): {permutaciones_directa}")
```
