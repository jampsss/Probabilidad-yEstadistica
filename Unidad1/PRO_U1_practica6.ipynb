```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar el Dataset
df = pd.read_csv('/content/autos2.csv')

# Limpieza inicial: convertir 'price' a numérico y eliminar nulos en esa columna
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df = df.dropna(subset=['price'])

display(df.head())
```

```python
# 2. Estadísticas descriptivas básicas y medidas de tendencia central del precio
min_p = df['price'].min()
max_p = df['price'].max()
moda_p = df['price'].mode()[0]
media_p = df['price'].mean()
mediana_p = df['price'].median()
std_p = df['price'].std()

print(f"--- Medidas de Tendencia Central y Dispersión ---")
print(f"Valor mínimo: {min_p}")
print(f"Valor máximo: {max_p}")
print(f"Moda: {moda_p}")
print(f"Media: {media_p:.2f}")
print(f"Mediana: {mediana_p}")
print(f"Desviación estándar: {std_p:.2f}")

print("\n--- Estadísticos y cuartiles ---")
display(df['price'].describe())
```

```python
# 3. Histograma de precios con Regla de Sturges
n = len(df['price'])
k = int(1 + np.log2(n)) # Regla de Sturges

plt.figure(figsize=(10, 6))
plt.hist(df['price'], bins=k, color='skyblue', edgecolor='black')
plt.title('Distribución de Precios de Automóviles (Regla de Sturges)')
plt.xlabel('Precio')
plt.ylabel('Frecuencia (Cantidad de Autos)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

### Interpretación del Histograma
La distribución de los precios presenta un sesgo a la derecha (asimetría positiva), lo que indica que la mayoría de los vehículos se encuentran en un rango de precio bajo a moderado, mientras que un grupo menor de vehículos alcanza precios considerablemente más altos (autos de lujo).

```python
# 4. Diagrama de caja: Precio vs Body Style
plt.figure(figsize=(12, 6))
sns.boxplot(x='body-style', y='price', data=df, hue='body-style', palette='Set2', legend=False)
plt.title('Relación: Precio vs Tipo de Vehículo (Body Style)')
plt.xlabel('Tipo de Vehículo')
plt.ylabel('Precio')
plt.show()
```

### Interpretación: Precio vs Body Style
Se observa que los vehículos tipo 'hardtop' y 'convertible' suelen tener los precios más altos y una mayor variabilidad. Por el contrario, los 'hatchback' presentan los precios más bajos y competitivos del mercado.

```python
# 5. Diagrama de caja: Precio vs Drive Wheels
plt.figure(figsize=(10, 6))
sns.boxplot(x='drive-wheels', y='price', data=df, hue='drive-wheels', palette='Pastel1', legend=False)
plt.title('Relación: Precio vs Tipo de Tracción (Drive Wheels)')
plt.xlabel('Tipo de Tracción')
plt.ylabel('Precio')
plt.show()
```

### Interpretación: Precio vs Drive Wheels
Los vehículos con tracción trasera (rwd) tienen un costo significativamente mayor en comparación con los de tracción delantera (fwd) o en las cuatro ruedas (4wd), lo cual suele asociarse a vehículos de mayor rendimiento o gama alta.

```python
# 6. Diagrama de caja: Precio vs Engine Location
plt.figure(figsize=(8, 6))
sns.boxplot(x='engine-location', y='price', data=df, hue='engine-location', palette='coolwarm', legend=False)
plt.title('Relación: Precio vs Ubicación del Motor')
plt.xlabel('Ubicación del Motor')
plt.ylabel('Precio')
plt.show()
```

### Interpretación: Precio vs Engine Location
Existe una diferencia drástica de precio basada en la ubicación del motor. Los vehículos con motor en la parte trasera ('rear') son mucho más costosos y exclusivos que aquellos con motor delantero ('front').
